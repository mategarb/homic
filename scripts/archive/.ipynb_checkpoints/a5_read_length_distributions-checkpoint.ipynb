```python
# define paths
path = "/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/Metagenomes"
db_path = "/gpfs/commons/home/mgarbulowski/016_proj_shm/ref_dbs/hs_wgs"

# samples id 
all_samps = ["KP005", "KP012", "KP013", "KP016", "KP024", "KP025", "KP026", "KP027", "KP029","KP033","KP035","KP038","KP040","KP046","KP047","KP048"]

## loading metadata
meta = pd.read_csv("/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/metadata_kp.csv")
meta = meta.loc[meta['id'].isin(all_samps)] # keep only the sequenced ones
```

```python
prc_gb=[]
prc_nt=[]
prc_rs=[]
prc_gbf=[]

for samp in all_samps:  
    path_blast_gb ="/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/blast_gbu_top5/" + samp + "_blastn_report.txt"
    path_blast_nt ="/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/blast_nt_top5/" + samp + "_blastn_report.txt"
    path_blast_rs ="/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/blast_rs_top5/" + samp + "_blastn_report.txt"
    path_blast_gbf ="/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/blast_gb_top5/" + samp + "_blastn_report.txt"
    
    path_ctgs = "/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/Metagenomes/metagenome_contigs_" + samp + ".fastq"

    prc_gb.append(process_data.perc_contigs_assigned(path_blast_gb, path_ctgs))
    prc_nt.append(process_data.perc_contigs_assigned(path_blast_nt, path_ctgs))
    prc_rs.append(process_data.perc_contigs_assigned(path_blast_rs, path_ctgs))
    prc_gbf.append(process_data.perc_contigs_assigned(path_blast_gbf, path_ctgs))
    
```

```python
print("Average amount of contigs assigned to species in gbbacu [%]: ")
print(round(statistics.mean(prc_gb) * 100,2))

print("Average amount of contigs assigned to species in nt [%]: ")
print(round(statistics.mean(prc_nt) * 100,2))

print("Average amount of contigs assigned to species in rs [%]: ")
print(round(statistics.mean(prc_rs) * 100,2))

print("Average amount of contigs assigned to species in gb [%]: ")
print(round(statistics.mean(prc_gbf) * 100,2))

db_nam = ["GbBacU","nt-blast","RefSeq","GeneBank"]
num_spec = [33604,102960590,1097218,1097235]


boxplot_style = {
    'medianprops': {'linewidth': 4, 'color': 'white'},
    'flierprops': {'marker': '*', 'markerfacecolor': 'black', 'markersize': 5},
}


# Example data
d = num_spec             # For bar plot
d_box = [prc_gb, prc_nt, prc_rs, prc_gbf]

colors = ['orangered', 'limegreen', 'dodgerblue', 'gold']
labels_bar = db_nam
labels_box = ['GbBacU', 'nt-blast', 'RefSeq','GeneBank']

fig, axes = plt.subplots(1, 2, figsize=(12, 6))  # 1 row, 2 columns

# --- First subplot: bar plot ---
ax = axes[0]
ax.bar(labels_bar, d, color=colors)
ax.set_ylabel('Number of species')
ax.set_yscale('log')

# --- Second subplot: boxplot ---
ax2 = axes[1]
bp = ax2.boxplot(d_box, patch_artist=True, notch=True, **boxplot_style)
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    
ax2.set_xticks([1, 2, 3, 4])
ax2.set_xticklabels(labels_box)
ax2.set_ylabel('Amount of contigs assigned to taxa with blastn')

axes[0].text(-0.1, 1.05, "A", transform=axes[0].transAxes,
        fontsize=16, fontweight='bold', va='top', ha='right')
axes[1].text(-0.1, 1.05, "B", transform=axes[1].transAxes,
        fontsize=16, fontweight='bold', va='top', ha='right')
plt.tight_layout()
plt.show()

fig.savefig("/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/figs/dbs_comparison.png", dpi=300, bbox_inches='tight')
```

```python
# saturation curve
reads_n = []
spec_n = []
chao1_all = []
for idx, samp in enumerate(all_samps):  
    reads_r1 = "/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/Metagenomes/016-Meta-" + str(idx+1) + "/" + samp + "_S" + str(idx+1) + "_L001_k2_1.fastq"
    reads_r2 = "/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/Metagenomes/016-Meta-" + str(idx+1) + "/" + samp + "_S" + str(idx+1) + "_L001_k2_2.fastq"
    
    tmp_r1 = file_readers.fastq(reads_r1)
    tmp_r2 = file_readers.fastq(reads_r2)
    reads_n.append(len(tmp_r1[0]) + len(tmp_r2[0]))
    
    path_blast = "/gpfs/commons/home/mgarbulowski/016_proj_shm/metagenomes_lib/blast_nt_top5/" + samp + "_blastn_report.txt"
    df = process_data.read_n_clean_blastn(path_blast, top_hits = True, drop_sp = True)
    chao1 = process_data.chao_idx(df["species"]) # m = 2, as pairwise 
    chao1_all.append(chao1[0])
    spec_n.append(df.shape[0])
    print(idx)
```
